```python
import pandas as pd;
```

```python
# Questions inside the casestudy drive.
# Task 1  - Read the Population dataset from the CSV file.
p = pd.read_csv("/content/drive/MyDrive/DSA - ICTAK/Case Study/Data Aquisition/population.csv");
print(p.head());
print(f"\nShape = {p.shape}");
print(f"\nDatatypes = {p.dtypes}");
```

```python
# Task 2 - Read the GDP dataset from the Excel file.
gdp = pd.read_excel("/content/drive/MyDrive/DSA - ICTAK/Case Study/Data Aquisition/gdp.xlsx");
print(gdp);
print(f"\nColumns = {gdp.columns}");
print(f"\nDatatypes = {gdp.dtypes}");
```

```python
# Task 3 - Read the Internet Users dataset from the JSON file.
json = pd.read_json("/content/drive/MyDrive/DSA - ICTAK/Case Study/Data Aquisition/internet_users.json");
print(json);
```

```python
# Task 4 - Read the Literacy Rate dataset from the XML file.
# Use xml.etree.ElementTree or pandas.read_xml().
xml = pd.read_xml("/content/drive/MyDrive/DSA - ICTAK/Case Study/Data Aquisition/literacy_rate.xml");
print(xml)
```

```python
# Task 5 - Explore each dataset.
# POPULATION DATASET.
print("\n-----POPULATION DATASET-----")
print(f"Number of rows and columns = {p.shape}");
print(f"\nColumn Names = {p.columns}");
print(f"\nDatatypes = {p.dtypes}");
print(f"\nMissing Values = {p.isnull().sum()}");
print(f"\nDuplicate = {p.duplicated().sum()}");

# GDP DATASET.
print("\n-----GDP DATASET-----")
print(f"Number of rows and columns = {gdp.shape}");
print(f"\nColumn Names = {gdp.columns}");
print(f"\nDatatypes = {gdp.dtypes}");
print(f"\nMissing Values = {gdp.isnull().sum()}");
print(f"\nDuplicate = {gdp.duplicated().sum()}")

# INTERNET DATASET.
print("\n-----INTERNET DATASET-----")
print(f"Number of rows and columns = {json.shape}");
print(f"\nColumn Names = {json.columns}");
print(f"\nDatatypes = {json.dtypes}");
print(f"\nMissing Values = {json.isnull().sum()}");
print(f"\nDuplicate = {json.duplicated().sum()}")

# LITERACY DATASET.
print("\n-----LITERACY DATASET-----");
print(f"Number of rows and columns = {xml.shape}");
print(f"\nColumn Names = {xml.columns}");
print(f"\nDatatypes = {xml.dtypes}");
print(f"\nMissing Values = {xml.isnull().sum()}");
print(f"\nDuplicate = {xml.duplicated().sum()}")
```

```python
# Task 6 - Rename columns wherever necessary so that all datasets have a common column name for merging.
gdp.rename(columns={"Country":"country"}, inplace = True)
print(gdp)
xml.rename(columns={"name":"country"}, inplace = True)
print(xml)
```

```python
# Task 7 - Convert the required columns into appropriate numeric data types.
# no changes needed, all datatypes are correct.
```

```python
# Task 8 - Merge all four datasets into one DataFrame.
df = pd.merge(p, gdp, on= "country");
df = pd.merge(df, json, on= "country");
df = pd.merge(df, xml, on= "country");
print(df)
```

```python
# Task 9
# Create the following calculated columns:

# 1. Internet Penetration (%)
df["Internet Penetration(%)"] = (df["internet_users"] / df["population"] * 100 ).round(2);

# 2. GDP Per Capita
df["GDP per capita"] = (df["GDP"]*100000 / df["population"]).round(2);
print(df)
```

```python
# Task 10
# Create a SQLite database. Store the final merged dataset into a table named ‘country_statistics’.
import sqlite3

con = sqlite3.connect("country_statistics.db");

df.to_sql("country_statistics", con, if_exists="replace", index=False);
```

```python
# Task 11
# Retrieve the first five records from the SQLite table..
result = pd.read_sql("SELECT * FROM country_statistics LIMIT 5", con)
print(result)
```

# **DATA ANALYSIS USING PANDAS**

```python
# 1 List the Top 5 countries by GDP.
print("Top 5 countries by GDP.")
print(df.nlargest(5,"GDP")[["country","GDP"]]);

# 2 List the Top 5 countries by population.
print("\nTop 5 countries by population.")
print(df.nlargest(5,"population")[["country","population"]]);

# 3 Calculate the average literacy rate.
print("\nAverage literacy rate.")
print(df["literacy_rate"].mean())

# 4 List all countries having a literacy rate greater than 90%.
print("\nLiteracy rate greater than 90%.")
print(df[df["literacy_rate"]>90][["country","literacy_rate"]])

# 5 List all countries having an internet penetration greater than 70%.
print("\nInternet penetration greater than 70%.")
print(df[df["Internet Penetration(%)"]>70][["country","Internet Penetration(%)"]])
```

```python
# 6 List the Top 5 countries by GDP per capita.
print("Countries by GDP per capita.");
print(df.nlargest(5,"GDP per capita")[["country","GDP per capita"]]);

# 7 Calculate the total population represented in the dataset.
print("\nTotal Population.");
print(df["population"].sum());

# 8 Calculate the average internet penetration across all countries.
print("\nAverage Internet Penetration");
print(df["Internet Penetration(%)"].mean());

# 9 Identify the country having the highest number of internet users.
print("\nHighest number of internet users.");
print(df.loc[df["internet_users"].idxmax(),["country","internet_users"]]);

# # 10 Find the correlation between Literacy Rate and Internet Penetration.
print("\nCorrelation between Literacy Rate and Internet Penetration.");
print(df["literacy_rate"].corr(df["Internet Penetration(%)"]));
```

# **Data Analysis Using SQL**

```python
# 11 Display the Top 5 countries by GDP.

q = """select country, GDP from country_statistics order by GDP DESC LIMIT 5"""
print("\nTop 5 countries by GDP.")
print(pd.read_sql(q,con))

# 12 Find the average literacy rate.
print("\nAverage literacy rate.")
q = """select AVG(literacy_rate) as average_literacy_rate from country_statistics"""
print(pd.read_sql(q,con))

# 13 Display all countries having literacy rate above 90%.
print("\nCountries having literacy rate above 90%.")
q = """select country, literacy_rate from country_statistics where literacy_rate>90"""
print(pd.read_sql(q,con))

# 14 Find the Top 5 countries with the highest internet penetration.
print("\nTop 5 countries with the highest internet penetration.")
q = """select country, "internet penetration(%)" from country_statistics
ORDER BY "internet Penetration(%)" DESC LIMIT 5"""
print(pd.read_sql(q, con))

# 15 Calculate the total population..
print("\nTotal population.")
q = """select sum(population) as total_popultion from country_statistics"""
print(pd.read_sql(q, con))

# 16 Find the country with the highest GDP per capita.
print("\nCountry with the highest GDP per capita.")
q = """select country, "GDP per capita" from country_statistics order by "GDP per capita" DESC LIMIT 1"""
print(pd.read_sql(q, con))

# 17 Export the final merged dataset as a CSV file.
df.to_csv("final_country_statistics.csv",index=False)
print("\nFinal CSV exported.")

# 18 Sort the dataset in descending order of GDP per Capita and display the Top 10
result = df.sort_values("GDP per capita", ascending = False)
print(result.head(10)[["country","GDP per capita"]])
# countries.
```
