```python
#1, 2, 3.
# Import pandas, numpy.
# Load dataset using pd.read_csv()  import numpy as np.

import pandas as pd
import numpy as np
file = "https://raw.githubusercontent.com/laxmimerit/All-CSV-ML-Data-Files-Download/master/IMDB-Movie-Data.csv"
df = pd.read_csv(file)

# Display: shape, head, column names, datatypes, missing value.

print("\nSHAPE")
print(df.shape)

print("\nHEAD")
print(df.head)

print("\nCOLUMN NAMES")
print(df.columns)

print("\nDATATYPES")
print(df.dtypes)

print("\nMISSING VALUE")
print(df.isnull())


```

```python
#4
# Check and remove duplicates.

print(df[df.duplicated()])

# Th e output shows there are no duplicate rows in the csv file.
# If there are duplicates, use df = df.drop_duplicates().
```

```python
#5
# Rename columns ("Title" > "Movie_Title").
df.rename(columns = {"Title":"Movie_Title"},inplace = True)

print(df.columns) # Shows name changed to Movie_Title.
```

```python
#6
# Get unique genres.
# Split the comma-separated string into a list of strings.
df["genre"] = df["Genre"].str.split(",")

# Explode the lists so each genre gets its own row, then find unique values.
unique_genres = df["genre"].explode().unique()
print(unique_genres)
```

```python
#7
# Extract number of genres per movie.
df["genre_count"] = df["Genre"].str.split(",").str.len()
print(df[["Movie_Title","genre_count"]])
```

```python
#8
# Handle the missing value using mean.
df["Revenue (Millions)"] = df["Revenue (Millions)"].fillna(df["Revenue (Millions)"].mean())

df["Metascore"] = df["Metascore"].fillna(df["Metascore"].mean())

# Count missing values in 'Revenue (Millions)' and 'Metascore.
print(f"Missing values in Revenue (Millions): {df["Revenue (Millions)"].isnull().sum()}")  #check how many missing values.

print(f"Missing values in Metascore : {df["Metascore"].isnull().sum()}")

```

```python
#9
# Top 5 movies by Rating.
print(df.nlargest(5,"Rating"))
```

```python
#10
# Movies with highest/lowest revenue.

print(f"Movie with Highest Revenue : {df.loc[df["Revenue (Millions)"].idxmax(),["Movie_Title"]]}")

print(f"Movie with Lowest Revenue : {df.loc[df["Revenue (Millions)"].idxmin(),["Movie_Title"]]}")
```

```python
#11
# Avg. rating by Director.

print(f"Rating by Director     {df.groupby("Director")["Rating"].mean()}")
```

```python
#12
# Movies per year.
print(df["Year"].value_counts())
```

```python
#13
# Movies with rating > 8 & revenue > 100M.
print(df.loc[(df["Rating"] > 8) & (df["Revenue (Millions)"] > 100), ["Movie_Title"]])
```

```python
#14
# Group by Director: avg Rating, Revenue.
print(df.groupby("Director")[["Rating", "Revenue (Millions)"]].mean())
```

```python
#15
# Group by Year: count, avg Rating.
print(df.groupby("Year").agg(Movie_count=("Year","size"),Average_Rating = ("Rating","mean")))
```

```python
#16
#  Create a new column Profitability:
# a) divided by Duration.
# b) Find top 10 most profitable movies.

df["Profitability"] = df["Revenue (Millions)"]/df["Runtime (Minutes)"]
print(df.columns)

print("\n\nTop 10 most profitable movies: \n")
print(df.nlargest(10, "Profitability")["Movie_Title"])
```

```python
#17
# Create a column that categorizes movies by rating:
# a) Rating < 6: Poor
# b) 6 ≤ Rating < 7.5: Average
# c) Rating ≥ 7.5: Excellent

categories = [
    df["Rating"] < 6,
    (df["Rating"] < 7.5),
    df["Rating"] >= 7.5
]

choices = ["Poor", "Average", "Excellent"]

df["Rating_Category"] = np.select(categories, choices, default="Undefined")
print(df[["Movie_Title","Rating", "Rating_Category"]])
```

```python
#18
#  Create a custom function to score movies based on:
# Apply row['Rating'] * 10 + row['Revenue (Millions)'] as High score
# Apply a function to dataset to assign the score

def movie_score(row):
    return row["Rating"] * 10 + row["Revenue (Millions)"]

df["Score"] = df.apply(movie_score, axis=1)

print(df[["Movie_Title","Score"]])
```
